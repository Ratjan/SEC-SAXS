```python
import os
import sys
import numpy as np
import matplotlib.pyplot as plt
import pandas as pd
```

## Stitches data from two detectors, the function concatenates the two files and sorts by q

```python
proposal = '20250420' # Only needed when working on MAXIV cloud
visit = '2025090908' # Only needed when working on MAXIV cloud

scans = ('62750',)  #list of scans to process

for scan in scans:
    EigdataFolder = f'/data/visitors/cosaxs/{proposal}/{visit}/process/scan-{scan}_shotsE'
    PildataFolder = f'/data/visitors/cosaxs/{proposal}/{visit}/process/scan-{scan}_shotsP'
    #EigdataFolder = f'input_stitch/scan-{scan}_shotsE' # For local testing
    #PildataFolder = f'input_stitch/scan-{scan}_shotsP' # For local testing

    if os.path.isdir(EigdataFolder):
        print(f'Folder exists:{EigdataFolder}')
        num_filesE= len([f for f in os.listdir(EigdataFolder) if 'shot_' in f])
        print(f'Number of files: {num_filesE}')
    else:
        print('Folder does not exist')

    if os.path.isdir(PildataFolder):
        print(f'Folder exists:{PildataFolder}')
        num_filesP= len([f for f in os.listdir(PildataFolder) if 'shot_' in f])
        print(f'Number of files: {num_filesP}')
    else:
        print('Folder does not exist')

    # Filter files containing "eiger" in their names
    files_eiger = sorted([f for f in os.listdir(EigdataFolder) if 'shot_' in f])
    files_pilatus = sorted([f for f in os.listdir(PildataFolder) if 'shot_' in f])

    # Print the filtered lists for verification
    #print("Eiger files:", files_eiger)
    #print("Pilatus files:", files_pilatus)

    # Check if files in both lists match
    for file_eiger, file_pilatus in zip(files_eiger, files_pilatus):
        if file_eiger[:6] != file_pilatus [:6]:
            print(f"Mismatch: {file_eiger} vs {file_pilatus}")

```

## Run time approx. 0.5-1 min/1800 frames

```python
from functions import stitch_data, save_stitched

for scan in scans:
    print(f'Stitching scan {scan}')
    for file_eiger, file_pilatus in zip(files_eiger, files_pilatus):
        df_stitched = stitch_data(EigdataFolder, PildataFolder, file_eiger, file_pilatus)
        save_stitched(df_stitched, file_eiger, scan, proposal, visit)
```
